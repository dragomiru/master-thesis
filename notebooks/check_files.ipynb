```python
import pandas as pd
import os
```

```python
df1 = pd.read_csv("../thesis_modularized/data/gemini1_pdf_processing_results.csv")
```

```python
df1_list = df1["pdf_name"].tolist()
```

```python
df1_list
```

```python
# List all PDF files in the directory
pdf_files = [f for f in os.listdir("../thesis_modularized/reports") if f.lower().endswith(".pdf")]

count = 0
df2_list = []
for file in pdf_files:
    df2_list.append(file)
```

```python
df2_list
```

```python
remaining_files = list(set(df2_list) - set(df1_list))
remaining_files
```

```python
source_dir = "../thesis_modularized/reports"
destination_dir = "../thesis_modularized/reports/remaining"
```

```python
import shutil

for filename in remaining_files:
    src_path = os.path.join(source_dir, filename)
    dst_path = os.path.join(destination_dir, filename)

    if os.path.exists(src_path):
        shutil.copy(src_path, dst_path)
        print(f"Copied: {filename}")
    else:
        print(f"File not found in source: {filename}")
```
